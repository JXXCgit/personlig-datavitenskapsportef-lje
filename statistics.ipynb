3. 数值型数据分布：
对称分布：比如正态分布
--正态分布：均数(u）和标准差sigma，N(u，sigma)，均数越大，越右移，标准差越大，曲线越矮胖（变异度参数）
--(u-sigma, u+sigma),置信区间对应68.27%，(u-1.96sigma, u+19.6sigma),置信区间对应95%，(u-2.58sigma, u+2.58sigma),置信区间对应99%
--标准正态分布：对正态分布进行标准化转化，Z=(X-u)/sigma, N(0,1)
偏态分布：偏左/正偏态分布，偏右/副偏态分布   
4. 数值变量的统计描述
--均数（总体均数：u,样本均数X-hat），用于正态分布
--中位数（M)，用于偏态分布资料，适合描述偏态分布的集中趋势
--离散趋势指标描述：
。。。极差/全距（R)，R=最大值-最小值
。。。方差(Variance)，表示一组变量值的平均离散程度，方差越大，离散程度越大，miu^2=sum(X-u)^2/N (总体方差)；miu^2=sum(X-u)^2/n-1 (样本方差)
。。。标准差（SD)， 表示一组变量组的平均离散程度，适用于正态分布。总体标准差=sqrt（miu^2）
。。。四分位数间距（Quartile),主要用于偏态分布资料， Q=P75-P25
。。。几何均数（适用于特殊的偏态分布资料）：对数正态分布资料，G=lg-1(sum(lgx)/n)
。。。变异系数（CV, coefficient of variation),用书描述数据的相对离散程度, CV=S/X-hat*100%,当多组数据的度量值单位不一样，或者数据的均数相差悬殊，只能用变异系数
总结：近似正态分布（均数，标准差）偏态分布（中位数，四分位数间距）
5. 分类变量的统计描述：
。。。相对数 （比，比例，率）
比：ratio，性别比，etc
比例：proportion，事物内部各部分所占的比重（构成比）
患病率（prevalence）：表示某时点接受检查的人群中现患病例所占的比例（是proportion），不是率
率：rate，表示单位时间内某事件发生的频率，分子：观察期间的某事件发生例数，分母：观察人-时总数（sum（每个观察单位x观察事件））
发病密度（incidence density）：观察期间内某疾病新发病例数/该观察其内的人-时总数
6. 两组数据变量比较的假设检验
。。。均值的抽样误差：抽样造成的样本均数与总体均数之间的差异（Sample error, SE),用标准误表示抽样误差，S（X-hat)=S/sqrt(n),增大样本量，降低抽样误差
。。。t分布：从正态分布N(u, sigma^2)总体中随机抽取样本量为n的样本，样本均数也服从正态分布N(u,sigma(X-hat)^2), Z=(X-hat-u)/sigma(X-hat)服从标准正态分布N(0,1)
t=(X-hat-u)/s(X-hat)=(X-hat-u)/s/sqrt(n)服从自由度n-1的t分布
t-分布是一组与自由度v有关的曲线，随着v的增大接近标准正态分布，对于曲线，中间部分包括95%的置信区间，那么两侧的界值称为0.05的t届值，
小概率原理：
前提条件H0：u=u0, t检验的|t|<t（0.025，v）, P>0.05, 接受H0
检验水准（显著性水准），alpha，通常取0.05
。。。 t检验方法：
单样本t检验（一个样本均数与已知总体均数u0的笔记，检验目的是推断未知总体均数u是否等于u0. 统计量t=（X-hat-u0)/(s/sqrt(n)),v=n-1 (自由度)，适用于正态性分布
两个样本独立t检验（推断u1是否等于u2），t=（X1-hat-X2-hat)/sqrt(S(X1-hat)-S(X2-hat)), v=n1+n2-2,适用于两个正态分布数据（或近似），两个样本独立，方差齐性（两个样本对应的总体方差相等，用levene方法检验）
配对样本t检验，常用于
1）对某些特征（性别，年龄）将两个受试对象配成对子，再对同一对子中的两个个体分别给予处理，每一对对子的两个观测变量为X,Y,差值为d=X-Y, d的样本均数对应的总体均数ud是未知的。检验的目的是推断ud是否等于0，统计量t=（d-hat-0)/sd/sqrt(n), v=n-1, n是对子数，效率高于两个样本的独立t检验
2）把同一受试对象分成两部分，再分别给予两种不同处理，
3）对同一受试对象处理前后的比较（自身对照设计）

p值不能反应总体均数差别大小，p值越小，越有把握两个样本均数不相等。
Z检验用于sigma未知，n较大的情况，但少用
假设检验的结论具有概率性：
HO原本正确，但P<=0.05,拒绝H0:第一类错误alpha
HO原本不正确，但P>0.05,接受H0:第二类错误beta
alpha为事先指定的检验水准，beta是未知的，alpha越小，beta越大。如果要同时降低这两类错误，需要增大样本量
7. Power: 检验效能或者把握度，是指当两总体缺失有差别时，按检验水准alpha，能够发现两总体间差别的能力，即1-beta，保证足够大的样本量，才能减少二类错误

############################ 统计
1. 20个高级统计方法
1.1 贝叶斯推断 (Bayes estimation)： PyMC库，
> 根据新的证据或信息更新假设的概率，允许你结合先验信念，并根据观察到的数据更新这些信念
1.2 最大似然估计（MLE)： scipy.stats
> 通过找到使得在某一统计模型下观测数据的似然最大化的参数值来进行估计，估计最适合数据的正态分布参数（均值、方差）
1.3 假设检验：t检验： scipy.stats.ttest_ind 
> 涉及到构造一个零假设（没有差异/效应）和一个备择假设。t 检验特别用于检验两个组的均值是否有显著差异。
1.4 方差分析：ANOVA：scipy.stats.f_oneway
> 检验三组或更多组的均值之间是否存在统计学上的显著差异
1.5 主成分分析（PCA)：sklearn.decomposition.PCA
> 将数据投影到新的正交轴（主成分）上来减少数据的维度，这些主成分能够捕捉到数据中的最大方差。
1.6 因子分析 （Factor analysis）：factor_analyzer 
> 将观察到的变量建模为潜在（未观察到）因子的线性组合，通常用于降维或揭示数据中的隐藏结构。
1.7 聚类分析 (K-means)： sklearn.cluster.KMeans
> 基于相似性划分为同质的组（簇）。K-means 是一种基于质心的常用聚类技术。
1.8 自助法 (booststrapping) 
> 通过从数据集中反复进行有放回的抽样，以估计统计量（如均值、中位数）的分布（和不确定性）。
1.9 时间序列分析（ARIMA，自回归积分)：statsmodels.tsa.arima.model
> 用于通过捕捉数据中的自相关性来预测单变量时间序列数据。
1.10 生存分析 （survival analysis）：lifelines库 import KaplanMeierFitter
> 处理时间到事件的数据，通常关注某一事件（如客户流失）发生的概率，尤其是在某个时间点之后。
1.11 多元回归（多元线性回归，multi-linear regression):sklearn.linear_model.LinearRegression
> 建模了一个因变量与多个自变量之间的关系。
1.12 
岭回归 （ridge regression）:sklearn.linear_model
> 在普通最小二乘回归的基础上增加L2正则化项，通过压缩回归系数来减少过拟合。
套索回归 （Lasso regression): sklearn.linear_model
> 增加L1正则化项，能够将一些回归系数压缩为零，从而实现特征选择。
1.13 逻辑回归（logistic regression）：sklearn.linear_model.LogisticRegression
> 用于二分类问题，建模某个类别或事件发生的概率。它通过Sigmoid函数将线性回归的输出转化为概率值，常用于分类任务。
1.14 混合效应模型（mixed effect model): statsmodels库
> 包含固定效应（对所有组通用的效应）和随机效应（特定于每个组的效应）。这种模型常用于处理纵向数据或分组数据。
1.15 非参数检验（mann-whitney U test): scipy.stats.mannwhitneyu
> Mann-Whitney U 检验用于比较两个独立样本是否来自同一分布，特别是当数据不符合正态分布时。
1.16 蒙特卡罗模拟 (Monte Carlo Simulation):
> 通过重复随机抽样来估计在不确定性下不同结果的概率。它是一种利用随机性解决问题的数学方法，广泛应用于风险分析、优化等领域。
1.17 马尔可夫链模特卡罗（MCMC)
> 在贝叶斯推断中用于从后验分布中生成样本，特别是在直接采样困难时。它们通过构造一个马尔可夫链，使得链的状态逐步逼近目标分布，广泛用于复杂的统计模型和高维参数估计。
参数估计，对于复杂的层次模型，直接积分无法解决时，使用 MCMC 进行采样来估计模型参数。
1.18 鲁棒回归（robust regression）：sklearn.linear_model.RANSACRegressor，
> 鲁棒回归方法（例如 RANSAC、Huber 回归）相比普通最小二乘回归，更不容易受到离群值的影响。它们通过在模型训练过程中减少离群值的权重或通过其他方法来减轻异常值的干扰，从而提高回归模型的稳健性。
1.19 copulas：copulas库
> 用于捕捉随机变量之间的依赖结构，而不涉及它们的边缘分布。Copulas 能够通过将边缘分布与依赖结构分开建模，描述变量之间复杂的关系。它们在金融领域尤其受欢迎，用于建模资产收益的联合分布。
1.20 广义加性模型（GAMs)：pyGAM库
> 通过允许预测变量使用非线性函数，同时保持加性特性，扩展了线性模型。与线性回归相比，GAMs 更具灵活性，但仍然保持可解释性，使其在许多实际问题中非常有用。

2. 统计指南
2.1 统计基础： 
> 数据类型：定性和定量 
> 描述统计：总结和描述数据集的主要特征：中心趋势的测量（均值，中值，众数，百分位数），离散度的测量（范围，方差，标准差）
> 推论统计：根据样本对总体做出预测或结论（假设检验，p值，置信区间）
2.2 概率论
> 概率分布：离散分布（泊松分布，二项式分布）， 连续分布（正态分布，指数分布）
> 中心极限定理：central limit theorem, 统计学的基石，它指出无论通体分布如何，样本均值的分布都会随着样本数量的增加而接近正态分布
> 贝叶斯统计：基于贝叶斯定理的统计推理方法，将先验知识与新数据结合，实时更新概率
2.3 统计建模
> 回归分析：检验变量之间的关系(线性回归，多元线性回归，逻辑回归)
> 时间序列分析：分析处理随时间收集的数据，旨在确定趋势、模式或季节变化。
> 方差分析（ANOVA）和多变量方差分析（MANOVA）:比较多组均值，anova专注单个因变量，MANOVA分析多个因变量
> 非参数方法：不依赖于特定分布假设，适用于非正态分布。Wilcoxon秩和检验、Kruskal-Wallis检验和Spearman相关系数。
2.4 机器学习和统计
> 监督学习， supervised learning,从被标记的数据中学习比如分类（如SVM、随机森林）和回归（如梯度提升）等任务。
> 无监督学习, unsupervised learning,从未标记的数据中学习比如集群（如K-means、DBSCAN）和降维（如PCA、t-SNE）等
> 深度学习，deep learning,主要负责具有多层的神经网络，能够学习复杂的模式和表示，比如图像识别、自然语言处理和强化学习。

2. 数据和抽样分布
总体 vs. 样本
2.1 随机抽样和样本偏差
总体：一个大型数据集，N表示总体规模
样本：总体的一个子集，n表示样本的规模
随机抽样：从总体中随机抽取元素到样本中
分层抽样：对总体分层，并在每层中做随机抽样
简单随机抽样：在不对总体分层的情况下，做随机抽样。抽样是可以有放回的，也可以是无放回的。
> 自选择抽样偏差（self-selection sampling)
样本偏差：样本对总体做出了错误的解释
2.2 选择偏差： 非随机抽样，主观随机挑选
> 趋均值回归：对同一变量连续测量时出现的现象，即在极端观测值后，会出现更趋向于中心的观测值。给极值给予特殊关注和意义，会导致某种形式的选择偏差。
2.3 统计量的抽样分布
> 中心极限定理：当样本的规模增大时，抽样分布呈正态分布的趋势
> 样本统计量：对抽取自大规模总体中的样本做计算得到的度量值
> 标准误差 （standard error）：多个样本间样本统计量的变异性,反映的是抽样估计的精确度（bias）。这与标准偏差（standard deviation，个体数据间的变异性（variance），反映数据的波动性）不一样
> 数据分布 vs. 抽样分布
通常，抽样分布比数据分布本身更加规则，分布更趋于正态分布，所以样本规模越大，样本统计量的分布就越窄
2.4 自助法: 从样本本身中有放回地抽取更多样本，并对每次重抽样重新计算统计量或模型，无需考虑抽样统计量符合正态分布
> 自助样本（bootstrap sample）：从观测数据集中做有放回的抽取而得到的样本
> 重抽样：在观测数据中重复抽取样本的过程，包括自助和置换（混洗/shuffel）
> bagging/bootstrap aggregating: 通过有放回地随机抽样生成多个训练集，并联独立训练多个模型，最后通过投票（分类）或者平均（回归）的方式合并结果，以减少模型的预测波动(降低variance）和过拟合，代表算法：random forest
> boosting: 通过串行（逐个）训练多个弱学习器，每个新模型专门纠正前一个模型的预测错误，最终把这些弱模型加权组合成一个高精度的强学习器，用于减少模型偏差（bias），提升模型的准确度。代表算法：XGBoost，LightGBM, GBDT, AdaBoost
2.5 置信区间：除了频数表，直方图，箱线图，标准误差等外，置信区间也可以用于了解一个样本估计量潜在的误差情况。
> 置信水平(CI)：由样本统计量所构造的总体参数的估计区间，比如，随机抽取1000人，测量身高平均值，标准误差为0.5， 正态分布的95%置信水平对应的临界值为+-1.96 （99%对应+-2.576，90%对应+-1.645），所以置信区间为170+-1.96*0.5=（169，171）。
2.6 正态分布：（连续型数据）X∼N(μ,σ2)
> 标准正态分布：均值为0，标准偏差为1
> 归一化（normalization）：为了和标准正态分布对比，将数据减去均值，然后除以标准偏差的过程，转化值为z分数。正态分布有时也叫z分布
> QQ图能可视化样本和正态分布间的近似度。通过对z分数从低到高排序，并将每个z分数绘制到y轴，x轴单位对应该值秩（rank）的正态分布所对应的分位数。如果数据大体落在对角线上，那么样本分布近似正态分布
2.7 长尾分布：（可以是离散型数据，也可以是连续型数据），如果正态分布mean=median，那么长尾分布数据：mean 》median （网络流量，财富，，，）
2.8 学生t分布: 成正态分布形状，但钟形稍厚，尾部略长，不知道总体标准差，样本量小（连续型数据）t=X-hat−μ0/(s/sqrt(n)), ​​σ 未知，用样本 s 估计,df=n−1
2.9 二项分布/伯努利分布：统计学的一种经典的离散概率分布，描述独立随机试验时，每次试验都具有相同概率情况下的总次数 （离散型数据）（100人中多少患病）,X∼Binomial(n, p)
2.10 泊松分布（poisson distribution) 及其相关分布: （离散型数据）
> 泊松分布 (poisson distribution)：固定一段时间，空间，范围内，某个事件发生多少次，只有一个参数：lambda。 （一天多少急诊病例）,X∼Poisson(lambda)
> 指数分布 (exponenial distribution): 等到下一次事件要多久，建模各次事件之间的时间分布情况（网络访问的时间间隔）,T∼Exponential(lambda)
> 韦伯分布（Weibull distribution）：一个事件多久发生 （机器寿命，生存时间）

| ----------- | --------------------------------------------------- |

3.3 重抽样：从观测数据中反复地抽取数据值，目标是评估一个统计量中的随机变异性 （resampling）
> 自助法（bootstrap) ：将现有样本视为总体的近似，通过从原样本中反复进行有放回抽样，来近似统计量的抽样分布，从而估计标准误，置信区间，以及其他不确定性 （通过有放回重抽样近似sampling distribution的方法），估计统计量的不确定性。
> 置换检验 (permutation test): 一种非参数的假设检验方法，通过在H0成立前提下，将各组数据合并后，按照原组大小进行无放回的随机重新分组，反复计算检验统计量，构造零假设下的参考分布（null distribution），并据此判断观察到的差异是否具有统计学显著性。
Permutation test — significance
H0 :no real difference between groups
Calculate observed difference: Δobs
Combine groups and randomly reassign labels many times.
Calculate Δperm each time.
These permutation differences form the null distribution:
the differences expected from random variation if H0 is true.
Compare Δobs with the null distribution. 
Δobs common under H0 → large p → not significant 
Δobs very rare/extreme under H0 → small p → significant
p<α⇒statistically significant

穷尽置换检验（exhaustive）：在零假设 H0 成立的前提下，将合并后的数据按照原来的组大小，枚举所有可能的重新分组方式，并计算每一种分组下的检验统计量，从而得到完整的 null distribution（无放回，所有组合都计算，仅适用于规模小的样本）
自助置换检验（bootstrap）：在置换检验的框架下，将各组数据合并后，按照原来的组大小进行有放回抽样来重新形成各组，反复计算检验统计量，以构造零假设下的参考分布

3.4 统计显著性和p值
> 用样本数据判断观察到的差异，是否大到不能合理地只用随机变异来解释，H0​:没有真实差异/没有真实效，如果 H0真的成立，我现在看到这么大的差异，常见吗？
> p值：假设 H0成立，观察到“当前这么大或更极端的差异”的概率
> α值：实验开始前人为设定的显著性阈值，也代表我们愿意接受的 Type I error 水平。
> 第一类错误：错误将一个有随机导致的效果归结为真
> 第二类错误：错误将一个为真的效果归结为随机导致的
3.5 t检验： 两个均值的差异，是否大到不能合理地用随机变异解释。
> 学生分布/t分布：总体标准差未知，用s代替σ， t分布是一个家族，形状取决于df（自由度），df越大，尾巴越薄，越接近正态分布 （正态分布95%对应的边界大约-1.96-1.96，t分布的95%对应的边界大约是-2.262-2.262. 
> t=(X-hat-u0)/(s/sqrt(n))=signal/noise, 观察的差异/差异的标准误
H0​:μ=μ0​,α=0.05,2.5% left tail+95% middle+2.5% right tail, ±tα/2,df​, if ∣t∣>tcritical​⇒p<α⇒reject H0​，else, ∣t∣≤tcritical​⇒p≥α⇒fail to reject H0​. 95%CI: X-hat±t_critical​×SE (t_critical:临界值)，如果 95% CI 不包含u0，对于双侧α=0.05 的检验，就对应：p<0.05，所以p-value 和 confidence interval 是同一个推断问题的两种表达
> one-sample t-test:比较一个样本均值与某个已知值。不同观察值是独立连续变量。如果是小样本，不应严重偏态或极端outlier。t=(X-hat-u0)/(s/sqrt(n))。H0​:μ=μ0​
> independent two-sample t-test：比较两组连续变量的均值是否存在统计学差异,两个总体方差相同,不同观察值是独立连续变量。t=(X_A-hat-X_B-hat)/(SE(X_A-hat-X_B-hat))=(X_A-hat-X_B-hat)/(sqrt(S_A^2/n_A+sqrt(S_B^2/n_B)))， 自由度：df=n_A+n_B−2。H0​:μA​−μB​=0
> welch's t-test：比较两个独立组，但不要求两组方差相等。H0​:μA​−μB​=0
> paired t-test：比较同一对象两次测量的平均差是否存在统计学差异, 不同 pair 之间相互独立。本质是对对 difference 做 one-sample t-test。 H0​:μD​=0，t=(D-hat-0)/(s_D/sqrt(n))​​,t=平均变化的不确定性/平均变化
summary: 
t-test assumptions
1. Independence
observations 应按研究设计保持独立
independent t-test：两组个体彼此独立
paired t-test：同一 pair 内可以相关，但不同 pair 之间应独立
避免 pseudoreplication
2. Outcome
主要用于连续变量
例如：blood pressure, weight, reaction time, revenue
3. Normality
不要求原始数据“完美正态”
小样本时，严重 skewness / extreme outliers 会更成问题
样本量较大时，t-test 对一定程度的非正态通常较 robust
paired t-test 关注的是：Di=Afteri−Beforei 的分布，而不是 Before 和 After 各自必须正态
4. Outliers
t-test 使用 mean 和 SD
mean 和 SD 都容易受极端值影响
所以 extreme outliers 往往比轻度 non-normality 更需要注意
5. Equal variance
只针对两个独立组比较。
Student's t-test:σA^2=σB^2, 假设两组总体方差相同。
Welch's t-test 不要求：σA^2=σB^2
实际分析中通常可以优先考虑 Welch's t-test。

Parametric: 假定一个具有特定参数结构的概率模型，对参数进行推断。

t-test
ANOVA
linear regression
t-test 主要关心：μ或：μA−μB

3.6 多重检验:同一个研究或同一批数据中同时进行多个假设检验，而随着检验次数增加，即使所有零假设都成立，也更容易仅由于随机变异得到一些 p<0.05 的假阳性结果。核心问题：检验次数越多，至少出现一个 Type I error（假阳性）的概率越高。至少一个假阳性的概率：P(≥1 false positive)=1−(1−α)^m,m 是检验次数
> FWER（Family-wise Error Rate）：FWER 是在一组假设检验中至少出现一个 Type I error 的概率。
> Bonferroni correction：Bonferroni 校正通过把每个检验的显著性阈值设为α∗=α/m, 来控制整体 FWER，但这种方法比较保守，可能降低 power、增加 Type II error。
> Holm correction：Holm 校正是一种逐步调整 p 值或显著性阈值的方法，它同样控制 FWER，但通常比 Bonferroni 更不保守、检验效能更高。
> FDR（False Discovery Rate）：FDR 是在所有被判定为显著的结果中，预期属于 false discoveries 的比例。
> Benjamini–Hochberg（BH）：Benjamini–Hochberg 是常见的 FDR 控制方法，特别适合基因组学、GWAS 或其他一次进行大量探索性检验的场景。
> FWER vs FDR：FWER 关注“这一组检验中是否出现任何一个假阳性”，而 FDR 关注“所有显著结果中有多大比例可能是假阳性”。
> ANOVA 与多重检验：当需要比较三个或更多组时，先做整体 ANOVA 可以避免直接进行大量两两 t 检验，但 ANOVA 显著后如果继续做组间比较，仍需要进行 multiple-comparison correction。
> A/B test 中的多重检验：如果一个 A/B test 同时检查 conversion、revenue、retention 等很多指标，每个指标都单独以 p<0.05 判断，会增加偶然发现显著结果的风险，因此通常需要提前确定 primary metric。
> p-hacking：不断尝试不同指标、subgroup、时间范围、排除标准或统计方法，直到找到 p<0.05，本质上也是一种未经控制的多重检验，会显著增加假阳性风险。
3.7 自由度:Degrees of Freedom (df),在已有约束条件下，一组数据中仍然可以独立变化的信息数量。因为均值由n 个 observation计算出来，会接近总体均值 μ，(xi​−x-hat)^2会比（xi-u)^2 稍微小一些，如果还直接除以n(而不是n-1),会系统性低估总体方差，n−1 既对应自由度，也起到了修正低估的作用，叫 Bessel's correction。
> One-sample t-test：df=n−1
> Paired t-test：df=npairs​−1
> Student independent two-sample t-test：df=nA​+nB​−2
> Welch's t-test 的自由度由两组的样本量和方差共同决定，可以不是整数，实际通常由软件计算。
> 自由度会影响 t 分布的形状：df↓⇒不确定性更大，t 分布尾部更厚,df↑⇒t 分布逐渐接近正态分布, 即使两个检验得到相同的 t 值，如果自由度不同，得到的 p-value 也可能不同
3.8 方差分析:ANOVA, Analysis of Variance,用于比较三个或以上组的总体均值是否相同的统计检验，它通过比较组间变异和组内变异来判断组别是否对结果变量产生了显著影响。H0​:μA​=μB​=μC​,H1​:至少有一组均值不同
>为什么不用很多个 t-test？做了多个 hypothesis tests，会增加：Type I error
>为什么叫“方差分析”？它不是直接看：X_A−X_B, 而是把数据中的 variation 拆成两部分: Between-group variation​, Within-group variation, 核心： F=between-group variation​​/within-group variation
> F≫1，说明不同组之间的差异明显大于组内随机波动，更可能得到：p<0.05
> ANOVA 显著后，通常还需要：post-hoc test （i.e., Tukey HSD)
> One-way ANOVA: 只有一个 factor
> Two-way ANOVA: 两个 factor, 可分析 a 主效应, b 主效应, a × b interaction
> ANOVA 的 assumptions和 t-test 很像，主要包括：
>> observations 的独立性
>> outcome 通常是连续变量
>> 各组数据不要有严重异常值
>> 经典 ANOVA 通常假设各组方差相近, 如果方差不齐，可以考虑：Welch ANOVA
3.9 卡方检验: Chi-square test, t-test / ANOVA 主要比较均值，而卡方检验主要比较分类数据的频数。 主要用来判断实际观察到的分类结果，和“如果没有差异/没有关系时应该出现的结果”相比，差得够不够大。
> χ^2=∑(O−E)^2/E,实际观察值和理论值差得有多远。
> Goodness-of-fit test （拟合优度检验），一个分类变量实际出现的比例，是否符合理论预期？
> Chi-square test of independence， 在 DS 和 A/B test 中更常见，问两个分类变量之间有没有关系？
> E=row total×column total/grand total
> χ^2↑, 意味着：Observed counts 和 Expected counts 差得越来越大。p↓. χ2越大⇒实际结果和 H0​预期越不一致​
> 卡方检验里的自由度: Goodness-of-fit中如果有：k个类别：df=k−1； Independence test中，如果 contingency table 有：r rows和：c columns， df=(r−1)(c−1)
> t-test→continuous outcome, χ^2→categorical outcome/counts
> 卡方检验有什么 assumptions: 数据是 counts,Observations 要独立, Expected count 不能太小, 至少约为 5,如果样本很小，考虑Fisher’s exact test​	​
> p<0.05, 只能说明两个 categorical variables 之间存在统计学关联，但不代表关联很大。和 t-test 一样：statistical significance ！= effect size
3.10 多臂老虎机算法： Multi-Armed Bandit, MAB，和 t 检验、ANOVA、卡方检验不太一样，因为它不是传统意义上的“显著性检验”，而是一个动态决策问题。一类在不断尝试不同选项的过程中，动态平衡“探索未知选项”和“利用当前最佳选项”，以最大化长期累计收益的方法。
> 核心矛盾：Exploration （我还不确定其他选项好不好，所以要继续尝试。） vs Exploitation（Exploitation=利用已经学到的信息赚钱/获得收益）
> A/B test (目标: inference, 通常固定分配流量并侧重统计推断)vs. multi-armed bandit (目标: reward maximization, 根据不断产生的数据动态调整各方案获得的流量​)
> Bandit 的主要目标通常是最大化累计收益, Learn and exploit simultaneously​，而传统 A/B test 的主要目标通常是获得清晰、可靠的 treatment-effect 推断, Learn first, exploit later
> Epsilon-Greedy: 大部分时间选择当前最佳选项，少部分时间随机探索其他选项, ϵ-greedy=(1−ϵ) exploit+ϵ explore
> Upper Confidence Bound: UCB=estimated performance+uncertainty bonus, ：同时考虑当前估计表现和估计的不确定性，对信息不足的选项给予探索机会
> Thompson Sampling: 利用 Bayesian posterior 表示每个选项效果的不确定性，并根据 posterior sampling 动态选择方案。
> Regret：由于没有选择真正最佳方案而损失的潜在收益，bandit 通常希望最小化 cumulative regret。
> Multi-armed bandit 常用于广告、推荐系统、网页内容和实时在线优化等可以快速获得用户反馈的场景。
3.11 检验效能(Power)和样本规模(Sample Size)
> power: 当真实效应确实存在并达到我们关心的大小时，统计检验能够正确检测出该效应的概率。
| 真实情况        | 检验结果：不显著      | 检验结果：显著      |
| ----------- | ------------- | ------------ |
| 实际没有 effect | 正确            | Type I error |
| 实际有 effect  | Type II error | 正确           |
α=P(Type I error), 实际没效果，却错误地认为有效。
β=P(Type II error),实际有真实 effect，但实验没检测出来。Power=1−β.比如：80% power： 如果真实 effect 达到了我们设计实验时指定的大小，那么这种实验有约 80% 的概率检测到它，约 20% 的概率漏掉它。Power 必须针对某个 effect size 来定义。
> 其他条件相同时：Effect size↑⇒Power↑， 反过来，想检测非常小的 effect：Power 不变的话，就必须增加：Sample size
> MDE（Minimum Detectable Effect）：实验设计时希望有足够统计效能检测到的最小效应大小。
> Effect size 描述效应有多大，而 MDE 是实验设计阶段选定的、希望能够可靠检测到的最小 effect size。 MDE=设计阶段, Observed effect=实验结束后
> Sample size=f(α, Power, MDE, Variability)​. 对于比例数据，还经常需要：baseline rate. α+Power+MDE+Variance/Baseline→n
> Power 越高，样本量越大, α↓⇒n↑, MDE↓⇒n↑​, Variance↑⇒n↑
> 样本也不能无限大. 因为样本量非常大时，即使非常微小、没有实际意义的 effect，也可能p<0.05.因此，不仅要看p值，还有effect size + confidence interval + practical significance
> Power analysis 是在实验开始前，根据显著性水平、目标 power、希望检测的 effect size 和数据 variability 来计算所需 sample size，或反过来评估已有 sample size 能达到多少 power。解决两个问题：我要达到 80% power，需要多少样本？我现在只能招 100 个样本，这个设计能达到多少 power？
> sample size的计算
对于连续变量： 两组均值比较
n=(2σ^2)*(z_1-α/2+z_1-β)^2/(MDE^2), 
σ,标准差，α=Type I error，1−β=Power， 
对于coversion rate类似的二元outcome：
n=[z_1-α*sqrt((2*p-hat)*(1-p-hat)) + (z_1-β)*sqrt(p1*(1-p1)+p2*(1-p2))]^2 /(p2-p1)*2,
p1​=A组 baseline conversion, p2​=B组希望检测到的 conversion, p-hat​=2p1​+p2, MDE=p2​−p1​​​
4. 回归与预测
4.1 简单线性回归: 用一个自变量 X 来描述或预测一个连续因变量 Y 与 X 之间线性关系的方法,Simple Linear Regression
> 样本估计Y^=b0+b1X, b1=3: X 每增加 1 个单位，Y 的预测平均值增加 3 个单位, b1=-1: X 每增加 1，预测 Y 平均减少 3, b0: intercept 主要只是数学上确定回归线位置的参数
> 完整的理论模型: Y=β0​+β1​X+ϵ, β0,β1代表总体中的真正参数, ϵ代表模型没有解释掉的随机变化。
> ei​=Yi​−Y^i​,每个 observation到回归线的垂直距离, 目标是让所有 residual 尽可能小。
> Least Squares: ∑ei^2​​=i=1∑n​(Yi​−Y^i​)^2, Ordinary Least Squares, OLS, 普通最小二乘法。
> b1​=∑(Xi​−Xˉ)(Yi​−Yˉ)​​​/∑(Xi​−Xˉ)^2, slope
> Pearson correlation：r,衡量X 和 Y 的线性关联强度。b1=r*(s_Y/s_X), cor(X,Y)=cor(Y,X), 
> Regression: X→Y, Correlation：关联有多强, Regression：X改变1单位，Y平均改变多少
> R^2: 模型解释了 Y 总变异的多少比例。R^2高,不等于：X 导致 Y。也不一定等于：模型预测一定很好。
> Regression ≠ causation
> Residual=Observed−Predicted​, SSE=i=1∑n(Yi−Yi-hat)^2 (Sum of Squared Errors)最小时，Ordinary Least Squares, OLS（普通最小二乘法）
> Least Squares = 选择使残差平方和最小的回归线
> Least Squares / OLS： 最小二乘法通过选择 b0和 b1，使所有 observation 的残差平方和最小，从而得到最佳拟合直线
> SST=sum(Y_i-Y-hat)^2, SSE=sum(Y_i-Y^_i​)^2，R^2=1-SSE/SST，这里，SST表示Y 原本总共有多少 variation，SSE表示模型还没有解释掉的 variation，但R^2高不等于模型一定好，并不能自动说明：X 导致 Y， 模型没有 bias， 对新数据预测一定好，regression assumptions 全部满足
> 如果主要目标是 prediction，之后还应该在 validation/test data 上看：MAE, RMSE, R^2
4.2 多元线性回归
> Y^=b0​+b1​X1​+b2​X2​+⋯+bp​Xp​​, 多元线性回归使用多个自变量共同解释或预测一个连续因变量，每个 coefficient 都是在“其他变量保持不变”时解释自己的X
> Coefficient in multiple regression： 在多元线性回归中，某个X_j的 coefficient 表示在其他模型变量保持不变时，X_j每增加 1 单位，预测 Y 的平均变化。
> 当不同 predictors 彼此相关时，简单回归中的 coefficient 可能混合其他变量的影响；加入其他 predictors 后，模型重新分配各变量与 Y 之间的关系。
> 简单回归看“总体关系”，多元回归看“控制其他变量后的关系”
> 多重共线性 Multicollinearity: ​如果两个 predictors 非常相似,它们可能高度相关。模型就会难以判断：Y 的变化到底应该归给X_1，还是归给X_2？结果可能：coefficient 不稳定，标准误变大， coefficient 对数据的小变化很敏感，甚至 coefficient 的方向发生变化，这不一定意味着整体预测一定很差。
4.3 使用回归做预测
> regression的一个目的是解释 X 和 Y 的关系，另外一个目的是prediction，Xnew​→model→Y^new​
> 训练数据预测得好，不代表新数据也预测得好,可能会overfitting。所以做prediction时，真正关心的是模型的generalization，模型对没见过的数据表现怎么样？
> MAE=1/n​∑∣Yi​−Y^i​∣， （Y^,预测值），表示平均预测错多少。
> RMSE=sqrt(sum(1/n​∑∣Yi​−Y^i​∣^2)),RMSE 对特别大的预测错误惩罚更重。
> R^2仍然可以用来描述模型解释了多少Y的variation，但在prediction场景中，最好在test data上看R^2
4.4 回归中的因子变量
> 如果回归中的因子变量是分类变量（Factor variable），必须转化为数字形式，才能进入回归模型。
> 分类变量放进回归，本质上就是把“属于哪个类别”变成几个 0/1 开关，让模型能计算。
> Dummy variables:用 0 和 1 表示某个 observation 是否属于某一类别的变量，使分类变量可以进入回归模型。比如，Bergen={0，1} 不​属于 Bergen/属于 Bergen​
> Reference category 是回归模型中的基准类别，其他类别的 coefficient 都表示相对于这个基准类别的差异,比如Region={Oslo,Bergen,Trondheim}，如果 Oslo 是 reference，那么：bBergen​表示 Bergen 相对于 Oslo 的差异；bTrondheim 表示 Trondheim 相对于 Oslo 的差异。
> 一个有 k 个类别的分类变量通常只需要 k-1 个 dummy variables，因为剩下的一个类别作为 reference category。	​
4.5 解释回归方程(Interpreting a Regression Equation):正确解释每个 coefficient
> Intercept 截距:当所有数值型 predictor 都等于 0，且所有分类变量都处于 reference category 时，模型预测的Y。
> Continuous predictor 的 coefficient: coefficient 表示在其他 predictors 保持不变时，该 predictor 每增加 1 个单位，预测 Y 的平均变化量。
> Categorical predictor 的 coefficient: coefficient 表示该类别相对于 reference category 的预测差异。
> coefficient 大，不代表变量更重要
> Interaction 交互作用: 表示一个 predictor 与Y的关系取决于另一个 predictor 的水平。
4.6 检验假设：回归诊断(Regression Diagnostics)
> Residual=Y−Y^,如果residual plot出现U型，弯曲，波浪形，说明模型没有捕捉到非线性关系，理想情况是residual在0上下随机分布，没有明显曲线结构
> Homoscedasticity: 残差的波动大小在不同预测值范围内大致保持稳定, 小预测值时 residual 和大预测值时也差不多
> Heteroscedasticity: 残差方差不恒定。小预测值时误差很小, 大预测值时误差越来越大
> Independence：线性回归通常假设 observations 之间的 errors 是独立的。
> autocorrelation: 如果residual error彼此相关
> Normality of residuals：残差是否近似正态, 常见检查方式：histogram of residuals, Q-Q plot, 如果 Q-Q plot 中点大致沿着直线：residual normality 比较合理。
> Outliers：异常值,某个 observation 的 Y 和模型预测值差得特别远。OLS 对 outliers 比较敏感
> Leverage：X 很极端
> Influential point：会明显改变模型的点,常见指标：Cook’s Distance
> Multicollinearity：多重共线性,如果两个 predictors 非常相关, 模型会比较难区分它们各自的作用。可能导致：coefficient 不稳定， standard error 变大， coefficient 对样本变化很敏感，常用检查VIF(Variation inflation factor)
> VIF 衡量由于 predictors 之间相关，某个 regression coefficient 的方差被放大了多少. VIF 越高，说明 multicollinearity 越严重。
4.7 多项式回归和样条回归（Polynomial Regression and Splines）
> Polynomial regression extends linear regression by including powers of a predictor, such as X^2 or X^3, to model nonlinear relationships.
> Y^=b0+b1X+b2X^2 (quadratic regression)
> Y^=b0+b1X+b2X^2 +b3X^3可以你和更复杂的曲线，但是degree越高，曲线越灵活，越容易拟合
> Splines 样条回归：： Spline regression models a nonlinear relationship by fitting low-degree polynomial functions over different ranges of X and joining them smoothly at knots.
> 高阶 polynomial 改一个地方，整条曲线都可能受影响， Spline某个区间的形状主要由局部数据决定， spline 更灵活，也更稳定
5. 分类: Regression 预测的是连续数值，Classification 预测的是类别。
5.1 朴素贝叶斯算法(Naive Bayes): probabilistic classification algorithm based on Bayes' theorem that assumes the predictors are conditionally independent given the class.
> P(A∣B)=P(B∣A)P(A)​/P(B), 在已经知道类别的情况下，各个 features 相互独立。假设3个features，X1=free, X2=prize, X3=click, 在已经知道类别的情况下，各个 features 相互独立,P(X1​,X2​,X3​∣Spam)=P(X1​∣Spam)P(X2​∣Spam)P(X3​∣Spam)
> Prior probability: the probability of a class before observing the current evidence,比如P(Spam)=0.2
> Posterior probability: the updated probability of a class after taking the observed evidence into account，比如P(Spam∣Words)=0.85
> 核心：根据features更新每个class的概率，然后选择概率最大的class
5.2 判别分析 (Discriminant Analysis): 根据每个类别的数据分布，判断新 observation 更可能属于哪个 class. 常见LDA：Linear Discriminant Analysis，QDA：Quadratic Discriminant Analysis
> LDA: 例如，Y={Healthy,Disease}， features：age， biomarker，在已经很多healthy和disease的训练数据的基础上，研究Healthy 这组的 X 大概分布在哪里？Disease 这组的 X 大概分布在哪里？然后然后来一个新病人：Age=70, Biomarker=8.2，模型判断这个人的 features 更像 Healthy 那一群，还是更像 Disease 那一群？
> LDA assumes class-specific multivariate normal distributions with a common covariance matrix, producing linear decision boundaries
> QDA: allows each class to have its own covariance matrix, producing more flexible curved decision boundaries.
> covariance matrix: matrix summarizes the variances of individual variables and the covariances between every pair of variables. ​描述两个变量是否一起变化。一方面表示每个变量自己散得多不多，一方面表示不同变量之间怎么一起变化
|               |               Age |         Biomarker |
| **Age**       |          Age自己的方差 | Age和Biomarker的协方差 |
| **Biomarker** | Biomarker和Age的协方差 |    Biomarker自己的方差 |
5.3 逻辑回顾 （Logistic Regression）
> 一种用于分类的方法，它根据一个或多个 predictors 估计 observation 属于某一类别的概率， X→P(Y=1)​
> logistic function/sigmoid function: p=1/(1+e^-z), z=b0​+b1​X1​+⋯+bp​Xp​
> Logistic function converts any real-valued linear combination of predictors into a probability between 0 and 1.
> X→linear combination→sigmoid→probability​->threshold-> class
> Odds: 发生概率和不发生概率的比值, odds=p/(1-p)=e^z, log(odds)=z=b0​+b1​X1​+⋯+bp​Xp
> odds ratio, OR: OR=odds_2/odds_1 =e^bj, 表示Xj每增加 1 个单位时，odds 乘以多少， OR>1：odds increase，OR<1：odds decrease，OR=1：no change in odds
> OR用于解释 logistic regression 的 coefficient，以及比较某个因素出现前后/不同组之间，事件发生的相对倾向有多大。例如： bsmoking​=1.10， OR=e^1.10≈3，在其他变量保持不变时，吸烟者患病的 odds 大约是不吸烟者的 3 倍
> Odds 描述一个状态；OR 比较两个状态。
5.4 评估分类模型
> | 真实情况       | 模型预测       | 名称                 |
| ---------- | ---------- | ------------------ |
| Disease    | Disease    | True Positive, TP  |
| No Disease | Disease    | False Positive, FP |
| No Disease | No Disease | True Negative, TN  |
| Disease    | No Disease | False Negative, FN |
> Accuracy: The proportion of all predictions that are correct. 所有人里面，模型一共预测对了多少？TP+TN. accuracy=(TP+TN)/(TP+TN+FP+FN)
> Precision: The proportion of predicted positives that are actually positive. 模型说“有病”的这些人里面，到底有多少真的有病？precision=TP/(TP+FP)
> Recall / Sensitivity: The proportion of actual positives that are correctly identified. 真正有病的人里面，我找出了多少？TP/(TP+FN)
> Specificity:The proportion of actual negatives that are correctly identified. 真正没病的人里面，我正确排除了多少？TN/(TN+FP)
> F1 Score: =2*(Precision×Recall)/(Precision+Recall)
> Threshold 会改变这些指标: p=0.5 -> p=0.3, TP↑, FP↑, Recall↑, Precision 或 Specificity 可能下降
> Precision:我说positive的人，有多少是真的？Recall: 真正positive的人，我找出了多少? Specificity:真正negative的人，我排除了多少？ 比如在医学上，如果是癌症筛选，漏掉真正病人FN的代价可能非常高。所以一般希望recall高一些，可能宁愿接受多一点FP。如果是一个很昂贵，很危险的治疗决定，FP的代价可能很高，一般更加重视precision
> ROC cure: 不同threshold下，看True Positive Rate （TP/(TP+FN)）和False Positive Rate （FPR=FP/(FP+TN）=1-specificality)
> ROC 曲线展示在不同分类阈值下，真正率 TPR 与假正率 FPR 之间的权衡。理想的模型希望TPR高，同时FPR低
> AUC (Area Under the ROC Curve): 范围 0-1，AUC=0.5表示和随机排序差不多。AUC=1表示完美区分positive和negative。AUC 越大，通常说明模型区分类别的能力越好。
> AUC高 !=某个实际threshold一定合适.还要考虑FP 的成本，FN 的成本，Recall，Precision，Specificity，实际 prevalence
> 在severe class imbalance下，Precision-Recall Curve很重要
> PR curve: 关注找到多少真正 positive？预测 positive 的人里面，有多少是真的？ 在 minority positive class 很重要时，PR curve 往往更有意义。比如fraud detection， rare disease detection，anomaly detection
> ROC看TPR vs FPR​，PR看Precision vs Recall，positive 和 negative 的整体 discrimination，positive class 找得怎么样。
