3. 数值型数据分布：
对称分布：比如正态分布
--正态分布：均数(u）和标准差sigma，N(u，sigma)，均数越大，越右移，标准差越大，曲线越矮胖（变异度参数）
--(u-sigma, u+sigma),置信区间对应68.27%，(u-1.96sigma, u+19.6sigma),置信区间对应95%，(u-2.58sigma, u+2.58sigma),置信区间对应99%
--标准正态分布：对正态分布进行标准化转化，Z=(X-u)/sigma, N(0,1)
偏态分布：偏左/正偏态分布，偏右/副偏态分布   
4. 数值变量的统计描述
--均数（总体均数：u,样本均数X-hat），用于正态分布
--中位数（M)，用于偏态分布资料，适合描述偏态分布的集中趋势
--离散趋势指标描述：
。。。极差/全距（R)，R=最大值-最小值
。。。方差(Variance)，表示一组变量值的平均离散程度，方差越大，离散程度越大，miu^2=sum(X-u)^2/N (总体方差)；miu^2=sum(X-u)^2/n-1 (样本方差)
。。。标准差（SD)， 表示一组变量组的平均离散程度，适用于正态分布。总体标准差=sqrt（miu^2）
。。。四分位数间距（Quartile),主要用于偏态分布资料， Q=P75-P25
。。。几何均数（适用于特殊的偏态分布资料）：对数正态分布资料，G=lg-1(sum(lgx)/n)
。。。变异系数（CV, coefficient of variation),用书描述数据的相对离散程度, CV=S/X-hat*100%,当多组数据的度量值单位不一样，或者数据的均数相差悬殊，只能用变异系数
总结：近似正态分布（均数，标准差）偏态分布（中位数，四分位数间距）
5. 分类变量的统计描述：
。。。相对数 （比，比例，率）
比：ratio，性别比，etc
比例：proportion，事物内部各部分所占的比重（构成比）
患病率（prevalence）：表示某时点接受检查的人群中现患病例所占的比例（是proportion），不是率
率：rate，表示单位时间内某事件发生的频率，分子：观察期间的某事件发生例数，分母：观察人-时总数（sum（每个观察单位x观察事件））
发病密度（incidence density）：观察期间内某疾病新发病例数/该观察其内的人-时总数
6. 两组数据变量比较的假设检验
。。。均值的抽样误差：抽样造成的样本均数与总体均数之间的差异（Sample error, SE),用标准误表示抽样误差，S（X-hat)=S/sqrt(n),增大样本量，降低抽样误差
。。。t分布：从正态分布N(u, sigma^2)总体中随机抽取样本量为n的样本，样本均数也服从正态分布N(u,sigma(X-hat)^2), Z=(X-hat-u)/sigma(X-hat)服从标准正态分布N(0,1)
t=(X-hat-u)/s(X-hat)=(X-hat-u)/s/sqrt(n)服从自由度n-1的t分布
t-分布是一组与自由度v有关的曲线，随着v的增大接近标准正态分布，对于曲线，中间部分包括95%的置信区间，那么两侧的界值称为0.05的t届值，
小概率原理：
前提条件H0：u=u0, t检验的|t|<t（0.025，v）, P>0.05, 接受H0
检验水准（显著性水准），alpha，通常取0.05
。。。 t检验方法：
单样本t检验（一个样本均数与已知总体均数u0的笔记，检验目的是推断未知总体均数u是否等于u0. 统计量t=（X-hat-u0)/(s/sqrt(n)),v=n-1 (自由度)，适用于正态性分布
两个样本独立t检验（推断u1是否等于u2），t=（X1-hat-X2-hat)/sqrt(S(X1-hat)-S(X2-hat)), v=n1+n2-2,适用于两个正态分布数据（或近似），两个样本独立，方差齐性（两个样本对应的总体方差相等，用levene方法检验）
配对样本t检验，常用于
1）对某些特征（性别，年龄）将两个受试对象配成对子，再对同一对子中的两个个体分别给予处理，每一对对子的两个观测变量为X,Y,差值为d=X-Y, d的样本均数对应的总体均数ud是未知的。检验的目的是推断ud是否等于0，统计量t=（d-hat-0)/sd/sqrt(n), v=n-1, n是对子数，效率高于两个样本的独立t检验
2）把同一受试对象分成两部分，再分别给予两种不同处理，
3）对同一受试对象处理前后的比较（自身对照设计）

p值不能反应总体均数差别大小，p值越小，越有把握两个样本均数不相等。
Z检验用于sigma未知，n较大的情况，但少用
假设检验的结论具有概率性：
HO原本正确，但P<=0.05,拒绝H0:第一类错误alpha
HO原本不正确，但P>0.05,接受H0:第二类错误beta
alpha为事先指定的检验水准，beta是未知的，alpha越小，beta越大。如果要同时降低这两类错误，需要增大样本量
7. Power: 检验效能或者把握度，是指当两总体缺失有差别时，按检验水准alpha，能够发现两总体间差别的能力，即1-beta，保证足够大的样本量，才能减少二类错误

############################ 统计
1. 20个高级统计方法
1.1 贝叶斯推断 (Bayes estimation)： PyMC库，
> 根据新的证据或信息更新假设的概率，允许你结合先验信念，并根据观察到的数据更新这些信念
1.2 最大似然估计（MLE)： scipy.stats
> 通过找到使得在某一统计模型下观测数据的似然最大化的参数值来进行估计，估计最适合数据的正态分布参数（均值、方差）
1.3 假设检验：t检验： scipy.stats.ttest_ind 
> 涉及到构造一个零假设（没有差异/效应）和一个备择假设。t 检验特别用于检验两个组的均值是否有显著差异。
1.4 方差分析：ANOVA：scipy.stats.f_oneway
> 检验三组或更多组的均值之间是否存在统计学上的显著差异
1.5 主成分分析（PCA)：sklearn.decomposition.PCA
> 将数据投影到新的正交轴（主成分）上来减少数据的维度，这些主成分能够捕捉到数据中的最大方差。
1.6 因子分析 （Factor analysis）：factor_analyzer 
> 将观察到的变量建模为潜在（未观察到）因子的线性组合，通常用于降维或揭示数据中的隐藏结构。
1.7 聚类分析 (K-means)： sklearn.cluster.KMeans
> 基于相似性划分为同质的组（簇）。K-means 是一种基于质心的常用聚类技术。
1.8 自助法 (booststrapping) 
> 通过从数据集中反复进行有放回的抽样，以估计统计量（如均值、中位数）的分布（和不确定性）。
1.9 时间序列分析（ARIMA，自回归积分)：statsmodels.tsa.arima.model
> 用于通过捕捉数据中的自相关性来预测单变量时间序列数据。
1.10 生存分析 （survival analysis）：lifelines库 import KaplanMeierFitter
> 处理时间到事件的数据，通常关注某一事件（如客户流失）发生的概率，尤其是在某个时间点之后。
1.11 多元回归（多元线性回归，multi-linear regression):sklearn.linear_model.LinearRegression
> 建模了一个因变量与多个自变量之间的关系。
1.12 
岭回归 （ridge regression）:sklearn.linear_model
> 在普通最小二乘回归的基础上增加L2正则化项，通过压缩回归系数来减少过拟合。
套索回归 （Lasso regression): sklearn.linear_model
> 增加L1正则化项，能够将一些回归系数压缩为零，从而实现特征选择。
1.13 逻辑回归（logistic regression）：sklearn.linear_model.LogisticRegression
> 用于二分类问题，建模某个类别或事件发生的概率。它通过Sigmoid函数将线性回归的输出转化为概率值，常用于分类任务。
1.14 混合效应模型（mixed effect model): statsmodels库
> 包含固定效应（对所有组通用的效应）和随机效应（特定于每个组的效应）。这种模型常用于处理纵向数据或分组数据。
1.15 非参数检验（mann-whitney U test): scipy.stats.mannwhitneyu
> Mann-Whitney U 检验用于比较两个独立样本是否来自同一分布，特别是当数据不符合正态分布时。
1.16 蒙特卡罗模拟 (Monte Carlo Simulation):
> 通过重复随机抽样来估计在不确定性下不同结果的概率。它是一种利用随机性解决问题的数学方法，广泛应用于风险分析、优化等领域。
1.17 马尔可夫链模特卡罗（MCMC)
> 在贝叶斯推断中用于从后验分布中生成样本，特别是在直接采样困难时。它们通过构造一个马尔可夫链，使得链的状态逐步逼近目标分布，广泛用于复杂的统计模型和高维参数估计。
参数估计，对于复杂的层次模型，直接积分无法解决时，使用 MCMC 进行采样来估计模型参数。
1.18 鲁棒回归（robust regression）：sklearn.linear_model.RANSACRegressor，
> 鲁棒回归方法（例如 RANSAC、Huber 回归）相比普通最小二乘回归，更不容易受到离群值的影响。它们通过在模型训练过程中减少离群值的权重或通过其他方法来减轻异常值的干扰，从而提高回归模型的稳健性。
1.19 copulas：copulas库
> 用于捕捉随机变量之间的依赖结构，而不涉及它们的边缘分布。Copulas 能够通过将边缘分布与依赖结构分开建模，描述变量之间复杂的关系。它们在金融领域尤其受欢迎，用于建模资产收益的联合分布。
1.20 广义加性模型（GAMs)：pyGAM库
> 通过允许预测变量使用非线性函数，同时保持加性特性，扩展了线性模型。与线性回归相比，GAMs 更具灵活性，但仍然保持可解释性，使其在许多实际问题中非常有用。

2. 统计指南
2.1 统计基础： 
> 数据类型：定性和定量 
> 描述统计：总结和描述数据集的主要特征：中心趋势的测量（均值，中值，众数，百分位数），离散度的测量（范围，方差，标准差）
> 推论统计：根据样本对总体做出预测或结论（假设检验，p值，置信区间）
2.2 概率论
> 概率分布：离散分布（泊松分布，二项式分布）， 连续分布（正态分布，指数分布）
> 中心极限定理：central limit theorem, 统计学的基石，它指出无论通体分布如何，样本均值的分布都会随着样本数量的增加而接近正态分布
> 贝叶斯统计：基于贝叶斯定理的统计推理方法，将先验知识与新数据结合，实时更新概率
2.3 统计建模
> 回归分析：检验变量之间的关系(线性回归，多元线性回归，逻辑回归)
> 时间序列分析：分析处理随时间收集的数据，旨在确定趋势、模式或季节变化。
> 方差分析（ANOVA）和多变量方差分析（MANOVA）:比较多组均值，anova专注单个因变量，MANOVA分析多个因变量
> 非参数方法：不依赖于特定分布假设，适用于非正态分布。Wilcoxon秩和检验、Kruskal-Wallis检验和Spearman相关系数。
2.4 机器学习和统计
> 监督学习， supervised learning,从被标记的数据中学习比如分类（如SVM、随机森林）和回归（如梯度提升）等任务。
> 无监督学习, unsupervised learning,从未标记的数据中学习比如集群（如K-means、DBSCAN）和降维（如PCA、t-SNE）等
> 深度学习，deep learning,主要负责具有多层的神经网络，能够学习复杂的模式和表示，比如图像识别、自然语言处理和强化学习。

两种基本类型：数值型 + 分类型
数值型：连续型（浮点数据，区间数据，数值数据），离散型（整数数据，技术数据，i.e., 次数）
分类型：二元数据（逻辑数据，布尔型数据），定序数据（如，数值评分）

3. 位置估计
均值
加权均值
中位数
百分位数/分位数：使得一定百分比的数据比它小的值
加权中位数：在排序数据中，使得它前面和后面的数据的权重之和是总权重一般的那个值
切尾均值：去掉一定数量的极端值值值后，所有值的平均数
健壮/耐抗性：Robust， 对极值不敏感
离群值/极端值：与多数数据截然不同的数据值
估计（统计学）：根据当前数据计算出某个值，以体现出从数据中得出的结果与理论上的真实结果之间的差别。（均值的替代估计）
度量（数据科学）：和估计一个概念

>>> 变异性估计 (variability/dispersion): 测量数据值是紧密聚集还是分散
> 偏差/误差/残差：观测值与位置估计之间的差异
> 方差/均方误差：与均值的偏差的平方和除以数据值的数量
> 标准误差：方差的平方根
> 平均绝对偏差/L1范数/曼哈顿范数：所有数据值与均值的偏差的绝对值的均值
> 中位数绝对偏差：所有数据值与中位数间偏差的绝对值的中位数
> 极差：数据集中最大值和最小值之差
> 顺序统计量/秩：基于从小到大排序的数据值的度量
> 百分位数：一个数据集集中百分之x的值小于等于它，百分之x的值大于等于它
> 分位数
四分位距（IQR)/四分卫差： 第75个百分位数和第25个百分位之间的差异
> 标准差以及相关估计：
