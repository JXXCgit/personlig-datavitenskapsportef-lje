1. 误差： 
抽样误差，随机测量误差 （不可避免）
系统误差，过失误差 （可避免）
2. 统计工作步骤：统计设计，搜集资料，整理资料，分析资料
3. 数值型数据分布：
对称分布：比如正态分布
--正态分布：均数(u）和标准差sigma，N(u，sigma)，均数越大，越右移，标准差越大，曲线越矮胖（变异度参数）
--(u-sigma, u+sigma),置信区间对应68.27%，(u-1.96sigma, u+19.6sigma),置信区间对应95%，(u-2.58sigma, u+2.58sigma),置信区间对应99%
--标准正态分布：对正态分布进行标准化转化，Z=(X-u)/sigma, N(0,1)
偏态分布：偏左/正偏态分布，偏右/副偏态分布   
4. 数值变量的统计描述
--均数（总体均数：u,样本均数X-hat），用于正态分布
--中位数（M)，用于偏态分布资料，适合描述偏态分布的集中趋势
--离散趋势指标描述：
。。。极差/全距（R)，R=最大值-最小值
。。。方差(Variance)，表示一组变量值的平均离散程度，方差越大，离散程度越大，miu^2=sum(X-u)^2/N (总体方差)；miu^2=sum(X-u)^2/n-1 (样本方差)
。。。标准差（SD)， 表示一组变量组的平均离散程度，适用于正态分布。总体标准差=sqrt（miu^2）
。。。四分位数间距（Quartile),主要用于偏态分布资料， Q=P75-P25
。。。几何均数（适用于特殊的偏态分布资料）：对数正态分布资料，G=lg-1(sum(lgx)/n)
。。。变异系数（CV, coefficient of variation),用书描述数据的相对离散程度, CV=S/X-hat*100%,当多组数据的度量值单位不一样，或者数据的均数相差悬殊，只能用变异系数
总结：近似正态分布（均数，标准差）偏态分布（中位数，四分位数间距）
5. 分类变量的统计描述：
。。。相对数 （比，比例，率）
比：ratio，性别比，etc
比例：proportion，事物内部各部分所占的比重（构成比）
患病率（prevalence）：表示某时点接受检查的人群中现患病例所占的比例（是proportion），不是率
率：rate，表示单位时间内某事件发生的频率，分子：观察期间的某事件发生例数，分母：观察人-时总数（sum（每个观察单位x观察事件））
发病密度（incidence density）：观察期间内某疾病新发病例数/该观察其内的人-时总数
6. 两组数据变量比较的假设检验
。。。均值的抽样误差：抽样造成的样本均数与总体均数之间的差异（Sample error, SE),用标准误表示抽样误差，S（X-hat)=S/sqrt(n),增大样本量，降低抽样误差
。。。t分布：从正态分布N(u, sigma^2)总体中随机抽取样本量为n的样本，样本均数也服从正态分布N(u,sigma(X-hat)^2), Z=(X-hat-u)/sigma(X-hat)服从标准正态分布N(0,1)
t=(X-hat-u)/s(X-hat)=(X-hat-u)/s/sqrt(n)服从自由度n-1的t分布
t-分布是一组与自由度v有关的曲线，随着v的增大接近标准正态分布，对于曲线，中间部分包括95%的置信区间，那么两侧的界值称为0.05的t届值，
小概率原理：
前提条件H0：u=u0, t检验的|t|<t（0.025，v）, P>0.05, 接受H0
检验水准（显著性水准），alpha，通常取0.05
。。。 t检验方法：
单样本t检验（一个样本均数与已知总体均数u0的笔记，检验目的是推断未知总体均数u是否等于u0. 统计量t=（X-hat-u0)/(s/sqrt(n)),v=n-1 (自由度)，适用于正态性分布
两个样本独立t检验（推断u1是否等于u2），t=（X1-hat-X2-hat)/sqrt(S(X1-hat)-S(X2-hat)), v=n1+n2-2,适用于两个正态分布数据（或近似），两个样本独立，方差齐性（两个样本对应的总体方差相等，用levene方法检验）
配对样本t检验，常用于
1）对某些特征（性别，年龄）将两个受试对象配成对子，再对同一对子中的两个个体分别给予处理，每一对对子的两个观测变量为X,Y,差值为d=X-Y, d的样本均数对应的总体均数ud是未知的。检验的目的是推断ud是否等于0，统计量t=（d-hat-0)/sd/sqrt(n), v=n-1, n是对子数，效率高于两个样本的独立t检验
2）把同一受试对象分成两部分，再分别给予两种不同处理，
3）对同一受试对象处理前后的比较（自身对照设计）

p值不能反应总体均数差别大小，p值越小，越有把握两个样本均数不相等。
Z检验用于sigma未知，n较大的情况，但少用
假设检验的结论具有概率性：
HO原本正确，但P<=0.05,拒绝H0:第一类错误alpha
HO原本不正确，但P>0.05,接受H0:第二类错误beta
alpha为事先指定的检验水准，beta是未知的，alpha越小，beta越大。如果要同时降低这两类错误，需要增大样本量
7. Power: 检验效能或者把握度，是指当两总体缺失有差别时，按检验水准alpha，能够发现两总体间差别的能力，即1-beta，保证足够大的样本量，才能减少二类错误
8. 参数统计和非参数统计：
参数统计：对样本所属的已知分布总体的未知参数进行估计或者假设检验，包括t-检验，方差分析，卡方检验，相关与回归。
非参数检验：直接比较分布，不拘于总体分布，将样本实际数据排序编秩次后，对秩次进行比较：
两个独立样本的秩和检验（wilcoxon两样本比较法）
配对设计差值的符号秩和检验（wilcoxon配对法）
